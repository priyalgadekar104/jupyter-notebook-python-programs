```python
#removing an item
s={10,20,30,40,50}
print("original set:", s)
item= int(input("enter item to remove:"))
if item in s:
    s.remove(item)
    print("updated set:", s)
else:
    print("item not found")
    
```

```python
a= {10,20,30,40}
b= {30,40,50,60}
c=a.intersection(b) #can also write c=a&b
print("set a:", a)
print("set b:", b)
print("intersection", c)

```

```python
a= {10,20,30,40}
b= {30,40,50,60}
c=a.union(b) #can also write a|b

print("set a:", a)
print("set b:", b)
print("union:", c)
```
