```python
#fill missing values
import pandas as pd
#define a dictionary with sample data which includes some missing values
data={
    'a':[1,2,3,None,5],
    'b':[None,2,3,4,5],
    'c':[1,2,None,None,5]
}
df= pd.DataFrame(data)
print("original data:\n",df)
#filling nan values with 0
df.fillna(0, inplace=True)
                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                              
```

```python
import pandas as pd
#define a dictionary with sample data which includes some midding values
data={
    'a':[1,2,3,None,5],
    'b':[None,2,3,4,5],
    'c':[1,2,None,None,5]
}
df=pd.DataFrame(data)
print("original data:\n",df)

#filling nan values with the mean of each column
df.fillna(df.mean(), inplace=True)
          
```

```python
import pandas as pd
data={
    'a':[1,2,2,3,3,4],
    'b':[5,6,6,7,8,8]
}
df=pd.DataFrame(data)
print("original DataFrame:\n", df.to_string(index=False))
#detect duplicates
print("duplicate rows:\n",df[df.duplicated()].to_string(index= False))
#remove duplicates based pn column'a'
df.drop_duplicates(subset=['a'], keep='first', inplace=True)
print ("DataFrame after removing duplicates based on column 'a' :\n",df.to_string(index=False))
```

```python
#rename column names to meaningful names
import pandas as pd
#sample data
data={
    'a':[25,30,35],
    'b':['john','doe','smith'],
    'c':[50000,60000,70000]
}
df=pd.DataFrame(data)

#rename columns
df.rename(columns={'a':'age', 'b':'name', 'c':'salary'},
          inplace=True)
print(df.to_string(index=False))
```
