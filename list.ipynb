```python
lst = []

for i in range(20):
    n = int(input("Enter value: "))
    lst.append(n)

# a) Count similar elements and print their index
print("\nIndex of each element:")
for i in range(len(lst)):
    print(lst[i], "-> index", i)

# b) Count even and odd values
even = 0
odd = 0

for n in lst:
    if n % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even values:", even)
print("Odd values:", odd)

# c) Count positive and negative values
positive = 0
negative = 0

for n in lst:
    if n > 0:
        positive += 1
    elif n < 0:
        negative += 1

print("Positive values:", positive)
print("Negative values:", negative)
```

```python
list1 = []

n = int(input("Enter number of elements in first list: "))
for i in range(n):
    list1.append(input("Enter element: "))

list2 = []

n = int(input("Enter number of elements in second list: "))
for i in range(n):
    list2.append(input("Enter element: "))

list3 = list1 + list2

print("First list:", list1)
print("Second list:", list2)
print("Merged list:", list3)
```

```python
lst=[]

while True:
    print("\n1. Insert")
    print("2. Delete")
    print("3. Display")
    print("4. Exit")

    choice=int(input("enter your choice:"))

    if choice==1:
        element= input("enter element:")
        position=int(input("enter position:"))
        lst.insert(position,element)
        print("element inserted.")

    elif choice==2:
        element=input("enter element to delete:")
        if element in lst:
            lst.remove(element)
            print("element deleted.")
        else: 
            print("element not found.")

    elif choice==3:
        print("list:", lst)

    elif choice==4:
        break 

    else:
        print("invalid choice")
```
